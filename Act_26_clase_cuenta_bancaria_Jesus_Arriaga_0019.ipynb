```python
class CuentaBancaria:
    def __init__(self, titular, numero_cuenta, saldo):
        self.titular = titular
        self.numero_cuenta = numero_cuenta
        self.saldo = saldo

    def leerdatos(self):
        self.titular = input("Nombre del titular: ")
        # Usamos input para mantener consistencia con tu ejemplo original
        self.numero_cuenta = input("Número de cuenta: ")
        self.saldo = input("Saldo actual: ")

    def obtener_formato(self):
        return f"Titular: {self.titular} | Cuenta: {self.numero_cuenta} | Saldo: ${self.saldo}"

    def mostrardatos(self):
        print(self.obtener_formato())

# --- Creación y uso de objetos ---

print("Datos de la Cuenta 1")
cuenta1 = CuentaBancaria("", "", 0)
cuenta1.leerdatos()
cuenta1.mostrardatos()

print("*_*_*_*_*_*_*_*_*_*_*__*_*_* ")

print("Datos de la Cuenta 2")
cuenta2 = CuentaBancaria("", "", 0)
cuenta2.leerdatos()
cuenta2.mostrardatos()

print("\nPrograma adaptado por Jesus Arriaga NL = 5")
```
